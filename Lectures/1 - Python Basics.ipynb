## Control flow

The most common control flow structures in Python are:
- `if` - conditional
    - `else` - used after `if` and runs when the condition is not met
    - `elif` - "else if", used after an `if` statement
- `for` - for loop
- `while` - while loop
- `break` - loop break
- `continue` - loop continue
- `pass` - do nothing

A __block__ in Python is defined by indentation. These can be spaces or tabs and must be _consistent throughout the entire code_. No mix of spaces and tabs!

__PEP 8 style guide__: Use spaces only (no tabs!) and 4 spaces at a time. VS Code editor will use four-space blocks by default. It also converts the tab to 4 spaces.

```python
# For loop with variable i ranging from 0 to 9, but stopping when i == 4

for i in range(10):
    if i == 4:  # Four spaces because the if block is inside the for block
        break   # Four more spaces (8 in total) because this is inside the if block.

# For loop with i ranging from 0 to 19 printing all even numbers
print("Even numbers between 0 and 20:")
for i in range(20):
    if i%2 != 0:
        continue  # Skip to next iteration
    print(f"  {i}")


# Using a while loop
i = 0
while i < 10:
    i += 1
print(f"While loop ran {i} times.")


# If-else/elseif blocks

a = 10
if a > 12:
    print(f"Greater than 12")
elif a > 3:
    print(f"Greater than 3")
else:
    print("3 or less")
```

### Function arguments and keyword arguments

Function arguments can be passed in the order they appear (position arguments), or explicitly named during the function call (keyword arguments).

Arguments may hold __default__ values.

```python
# Define a new function with c holding a default value.
def new_function(a, b, c=3):
    return a+b*c


print(new_function(10, 20))  # Position argument call

print(new_function(10, b=3, c=4))  # Keyword argument (assigned by name)

print(new_function(b=3, c=4, a=5))  # Keyword argument in a different order

print(new_function('d', 'ef'))
```

## Indexing and slicing

Slicing is a technique in Python used to extract segments of a sequence (such as lists, strings, tuples, etc.). It creates new lists from existing ones by proving the starting and ending indices. The starting index is inclusive, the ending index is not, meaning the slice will not contain the element at the last position.

The syntax for slicing is: `sequence[start:end:step]`. Where `start` is the starting index, `end` is the ending index, and `step` is the size of the step (defaults to 1).

```python
l = list(range(20))

l_slice = l[5:10]  # Slice list l from index 5 to 10

print(l_slice)
```

### Sets

Sets in Python are unordered collections designed to perform efficient membership lookup. Sets do not accept duplicates, therefore, each element in a set is unique. Element duplicates are simply ignored.

Common uses of sets:
- Removing duplicates from a list
- Perform quick membership lookups
- Find common elements between two collections

```python
# Removing duplicates from a list

numbers = [1, 2, 2, 3, 3, 4, 5, 5]
unique_numbers = set(numbers)  # Convert list numbers into a set
print(f"List: {numbers}")
print(f"Duplicates removed: {unique_numbers}")
```

```python
# Fast membership testing
fruits = {"apple", "banana", "orange", "strawberry"}
print("orange" in fruits)  # This should print True
print("grape" in fruits)  # This should print False
```

```python
# Set operations

a = {1, 2, 3}
b = {3, 4, 5}

print(f"Intersection: {a & b}")
print(f"Union: {a | b}")
print(f"Difference: {a - b}")
```

### Dictionaries

Dictionaries combine the lookup efficiency of sets with the indexing capabilities of lists. Dictionaries store __key-value__ pairs. Keys are used as index, it is the parameter you input when looking for something. Values hold the information to be stored/retrieved.

```python
# Creating a dictionary

person = {
    "name": "Alice",
    "age": 30,
    "city": "Philadelphia"
}

# Access dict fields
print(f"Name: {person['name']}")
print(f"Age: {person['age']}")
print(f"City: {person['city']}")


# Updating fields
person['age'] = 34
print(f"Age updated: {person['age']}")
```

### String search and check

You can search substrings within a string using the following methods:

- `.find(sub)/rfind(sub)` - find the first/last occurrence of a substring. Returns `-1` if not found.
- `.startswith(sub)/.endswith(sub)` - check if the string starts with or ends with substring `sub`.
- `.count(sub)` - count how many times `sub` appears in a substring.

```python
## Find substrings

s = "Hello world"
print(s.find("o"))  # Find first
print(s.rfind("o"))  # Find last
```

```python
## Startswith/endwith

print(s.startswith("He"))
print(s.endswith("rld"))
print(s.startswith("he"))  # Case sensitive!
```

## Pythonic idioms

Python has a number of features that permit performing complex computations with few lines of code (or even one-liners).

### Ternary operator

The ternary operator in Python performs conditional checks and assign values based on the result of such checks in a single line.


The syntax is: `<True_result> if <True_condition> else <False_result>`.

Consider the program below that checks whether a number `n` is _even_ or _odd.

```python
# Check even or odd

n = 10

if n % 2 == 0:
    result = "even"
else:
    result = "odd"

print(f"n is {result}")
```

Now, consider the exerpt below that does the same but in one line.

```python
# Checking even or odd with the ternary operator

# result = "even" if n % 2 == 0 else "odd"
# print(f"n is {result}")
n = 3
print("n is even" if n % 2 == 0 else "n is odd")
```

### Comprehensions

Python comprehensions are used to quickly create and initialize data structures in a single line of code. We will cover:

- List comprehensions: `[x for x in L]` 
- Set comprehensions: `{x for x in L}` 
- Dictionary comprehensions `{x: y for x, y in L}`

In the above examples, `L` is an iterable (like a list, set, etc.).

```python
# Python comprehensions

# Create a list with the squares of numbers from 0 to 9
squares = [i**2 for i in range(10)]
```

```python
# Using conditionals to tilter the above list to inclue only even numbers
even_squares = [x for x in squares if x % 2 == 0]
```

```python
# Set comprehensions work similarly to list comprehensions but we use { } instead of [ ]

set_squares = {i**2 for i in range(10)}
```

```python
# Use list comprehension to flatten a 2D list (matrix)

matrix = [[1,2], [3,4], [5,6]]
flattened = [x for row in matrix for x in row]

print(matrix)
print(flattened)
```

## Zipping, enumerating and unpacking

You can use `zip` to iterate simultaneously over multiple lists (or iterables). _Note_: if both lists are of different sizes, `zip` will discard the remainder of the longest list.

Example:

```python
## Create two lists

t1 = list(range(10))  # List numbers from 0 to 9
t2 = list(range(10, 20))  # List numbers from 10 to 19

# Here, the for loop uses two iterators `a`` and `b` - one for each list
for a, b in zip(t1, t2):
    print(f"{a} is in t1 | {b} is in t2")
```

### Lambda functions

Lambda functions are _annonymous_ functions defined inline on your program used to evaluate one or more arguments into a single expression.

The syntax is `lambda x: <expression>` where `x` is an input parameter to the function and `<expression>` is any valid Python expression that will produce a return value.

```python
# Defining a lambda function that returns the square of a number

f = lambda x: x**2

print(f(2))
```

## The %timeit magic function


Python has a _magic_ function for measuring the running time of programs named `%timeit`.

```python
%%timeit
# Notice the double '%'. This means %timeit will run over the entire program


l = list()

for i in range(100000):
    l.append(i**2)
```

```python
%%timeit
# Notice the double '%'. This means %timeit will run over the entire program



l = [i**2 for i in range(100000)]
```

```python
# Exercise

# Create a large list and a large set using comprehensions containing numbers from 0 to 1000000
# Use the membership operator `in` to find the element `999999` in each list
# Measure the running times with %%timeit


large_list = list(range(1000000))
large_set = set(range(1000000))

target = 999999

print("- Measuring search time using `list`")
%timeit target in large_list


print("- Measuring search time using `set`")
%timeit target in large_set
```
