# The 20newsgroups dataset

The dataset to be used in this lab is called `20newsgroups`. It can be fetched using Scikit-Learn's `fetch_20newsgroups` functions.

It consists of news articles from different categories, including science, medicine, religion, computers, and more.

The below snippet loads the dataset by filtering a few categories:

````python
from sklearn.datasets import fetch_20newsgroups

categories = ['alt.atheism', 'comp.graphics', 'sci.med', 'soc.religion.christian']
data = fetch_20newsgroups(subset='all', categories=categories, remove=('headers', 'footers', 'quotes'))
````

### 1a. Splitting the dataset

- Use the `train_test_split` function to split the data into four parts: `X_train`, `X_test`, `y_train`, `y_test` with `test_size=0.4`.
- Print the first 4 elements in `X_train`.

Note that `X_train` and `X_test` are _not feature matrices_. They are lists of strings that still must be converted to a feature matrix.


````python
# 1a.
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.4)
````

### 1b.

Create a feature matrix `X` using `CountVectorizer()` and print its shape.

````python
# 1b.

from sklearn.feature_extraction.text import CountVectorizer

X = CountVectorizer().fit(X_train)
````

### 1c.

Create multiple pipelines using `make_pipeline`, combining different ways to vectorize text data with different classification models.

For the text vectorizers, create pipelines with `CountVectorizer`, `TfidfVectorizer` and `SentenceTransformerEmbedding`. For classifiers, use `KNeighborsClassifier` with `n=3` neighbors and `SVC`.

In total, you will need 6 different pipelines such as (`CountVectorizer`, `KNeighborsClassifier`), (`TfidfVectorizer`, `KNeighborsClassifier`),...(`SentenceTransformerEmbedding`, `SVC`).

````python
# 1c.
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn_embeddings import SentenceTransformerEmbedding
from sklearn.pipeline import make_pipeline


vectorizers = {
    "CountVectorizer": CountVectorizer(),
    "TfidfVectorizer": TfidfVectorizer(),
    "SentenceEmbedding": SentenceTransformerEmbedding()
}

classifiers = {
    "KNeighborsClassifier": KNeighborsClassifier(n_neighbors=3),
    "SVM": SVC()
}

# Create the pipelines combining the above vectorizers and classifiers
# Suggestion: Create the pipelines as a dictionary of name->pipeline where `name` describes the vectorizer and classifier used.
pipelines = {
    "count-kn": make_pipeline(CountVectorizer(), KNeighborsClassifier()),
    "count-svm": make_pipeline(CountVectorizer(), SVC()),
    "tfidf-kn": make_pipeline(TfidfVectorizer(), KNeighborsClassifier(n_neighbors=3)),
    "tfidf-svm": make_pipeline(TfidfVectorizer(), SVC()),
    "emb-kn": make_pipeline(SentenceTransformerEmbedding(), KNeighborsClassifier()),
    "emb-svm": make_pipeline(SentenceTransformerEmbedding(), SVC())
}


````

### 1d.

Train each pipeline on the training set, and test them on the test set. Use `accuracy_score` to measure the model's accuracy.

````python
from sklearn.metrics import accuracy_score

for name, p in pipelines.items():
    model = p.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(name, acc)
````
